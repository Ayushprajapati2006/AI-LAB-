```python
print("Ayush Prajapati -1WA24CS075")
import heapq

def misplaced_tiles(state, goal):
    count = 0
    for r in range(3):
        for c in range(3):
            if state[r][c] != 0 and state[r][c] != goal[r][c]:
                count += 1
    return count

def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    neighbors = []
    r, c = find_blank(state)
    moves = [
        (0, -1, "Left"),
        (-1, 0, "Up"),
        (0, 1, "Right"),
        (1, 0, "Down")
    ]
    for dr, dc, move_name in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_board = [list(row) for row in state]
            new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
            neighbors.append((tuple(tuple(row) for row in new_board), move_name))
    return neighbors

def print_matrix(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))

def solve_a_star(initial, goal):
    # Priority queue stores: (f, g, current_state, path_records)
    # path_records stores tuples of (state, move_name, h)
    h_init = misplaced_tiles(initial, goal)
    queue = [(0 + h_init, 0, initial, [(initial, "Initial", h_init)])]
    visited = set()

    while queue:
        f, g, current_state, path = heapq.heappop(queue)

        if current_state in visited:
            continue
        visited.add(current_state)

        if current_state == goal:
            return path

        for neighbor, move in get_neighbors(current_state):
            if neighbor not in visited:
                h = misplaced_tiles(neighbor, goal)
                next_f = (g + 1) + h
                queue.append((next_f, g + 1, neighbor, path + [(neighbor, move, h)]))
    return None

# --- Configuration matching the 2nd image ---
INITIAL_STATE = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

GOAL_STATE = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

solution = solve_a_star(INITIAL_STATE, GOAL_STATE)

if solution:
    for g, (state, move, h) in enumerate(solution):
        print(f"Move: {move}")
        print(f"g = {g}")
        print(f"h = {h}")
        print(f"f = {g + h}")
        print_matrix(state)
        print("-" * 20)

```
