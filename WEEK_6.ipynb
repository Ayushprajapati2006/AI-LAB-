```python
# -*- coding: utf-8 -*-
"""1WA24CS083_Week6_PropositionalLogic_TruthTableEnumeration.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/1Y8XN8hRDhYODbF9ytkXz0KbiOlb2OCNZ
"""

from itertools import product

def implies(a, b):
    return not a or b

def KB(p, q, r):
    return implies(q, p) and implies(p, not q) and (q or r)

def entails(query):
    for p, q, r in product([False, True], repeat=3):
        if KB(p, q, r) and not query(p, q, r):
            return False
    return True

while True:
    print("\n1. R")
    print("2. R -> P")
    print("3. Q -> R")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == '4':
        print("Program terminated.")
        break

    if choice == '1':
        query = lambda p, q, r: r
    elif choice == '2':
        query = lambda p, q, r: implies(r, p)
    elif choice == '3':
        query = lambda p, q, r: implies(q, r)
    else:
        print("Invalid choice! Try again.")
        continue

    if entails(query):
        print("KB entails the query")
    else:
        print("KB does not entail the query")

print("Ayush Prajapati 1WA24CS075")
```
